Notebook with some plots to show correlations of the features with respect to being in the top 20.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
import gzip
import numpy as np
```

```python
# Take the merged dataset cleaned from null values and anomalies
with gzip.open('../../dataset/train_set.pkl', 'r') as file:
    dataset = pickle.load(file)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from mpl_toolkits.mplot3d import Axes3D

features = ['avg_position', 'avg_position_1', 'avg_position_2', 'avg_position_3', 'avg_position_4']

# Creare lo scatter plot
plt.figure(figsize=(8, 6))

sampled_data = dataset.sample(frac=0.1, random_state=42)  # Prendi il 10% dei dati

# Creare una griglia di scatterplot usando tutte le combinazioni di 'features'
sns.pairplot(sampled_data, vars=features, hue='top20', palette='coolwarm', plot_kws={'s': 10, 'edgecolor': 'k', 'alpha': 0.6})
plt.show()
```

```python
features2 = ['performance_wma','top20_consistency','races_dby_consistency']

# Creare una griglia di scatterplot usando tutte le combinazioni di 'features'
sns.pairplot(sampled_data, vars=features2, hue='top20', palette='coolwarm', plot_kws={'s': 10, 'edgecolor': 'k', 'alpha': 0.6})
plt.show()
```
