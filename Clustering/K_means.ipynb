# Cyclist cluster - KMeans
### K = 3
### DROP ADJUSTED_POINT < 271
### FEATURES: 'adjusted_points', 'avg_position_profile_1', 'avg_position_profile_2','avg_position_profile_3', 'avg_position_profile_4', 'avg_position_profile_5'

In this clustering analysis, we applied a filter to include only cyclists with adjusted_points greater than 271. This step was taken to reduce interference from cyclists with low participation or inconsistent performances in races. By doing so, the analysis focuses on more relevant data and higher-performing cyclists.

With this refined dataset, the optimal number of clusters was reduced to K = 3, simplifying the structure and providing clearer group differentiation. Here’s what the clusters reveal:
  - Cluster (Blue): Cyclists with high adjusted_points and strong average positions across all race profiles. This group represents top performers who are consistent across different types of races, finally, the results were the same as the previous.
  - Cluster (Orange): Cyclists with moderate adjusted_points and solid but slightly less consistent average positions. These cyclists perform well but may not dominate across all profiles.
  - Cluster (Green): Cyclists with lower adjusted_points (still above the 271 threshold) and more variability in their average positions. This group includes capable but less consistent cyclists.

Filtering cyclists with adjusted_points > 271 improved the clarity of the clusters by removing outliers with low participation and performance. This refined approach provides a cleaner and more focused analysis, revealing how high-performing cyclists distribute across race profiles and their consistency levels.


```python
# Load dataset
with gzip.open('dataset/cyclist_nf.pkl', 'r') as file:
    cyclists = pkl.load(file)
# Drop the rords with adjiusted_points <= 271
cyclists = cyclists[cyclists['adjusted_points'] > 271]
# Features list to use
features = ['adjusted_points', 'avg_position_profile_1', 'avg_position_profile_2','avg_position_profile_3', 'avg_position_profile_4', 'avg_position_profile_5']
ut.clustering_K_means(cyclists, features, 15, 'Cyclist', False, True)
```
