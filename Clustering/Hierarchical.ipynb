# Cyclist cluster - Hierarchical
### K = 3
### FEATURES: 'avg_position_profile_1', 'avg_position_profile_2', 'avg_position_profile_3', 'avg_position_profile_4', 'avg_position_profile_5'

We applied hierarchical clustering to group cyclists based on their average positions across five race profiles. Unlike previous methods (e.g., K-means or X-means), hierarchical clustering builds clusters step-by-step, providing a tree-like structure that allows for flexible decisions on the number of clusters.

The method resulted in 3 distinct clusters:
  - One cluster identifies high-performing cyclists with the best average positions across all race profiles. These cyclists consistently achieve top results, showing their dominance across flat, hilly, and mountainous terrains.
  - A second cluster captures mid-level performers who achieve moderate results. While they maintain decent average positions, they lack the consistency or excellence of the top-performing group.
  - The third cluster groups low-performing cyclists with weaker average positions across all profiles, indicating less competitive results overall.

Compared to K-means and X-means, hierarchical clustering maintains the same overall structure, with the top-performing group remaining stable. However, it clearly separates the lower and mid-level performers without requiring predefined cluster numbers, offering a more interpretable and flexible solution.

```python
# Load the dataset
with gzip.open('../dataset/cyclist_nf.pkl', 'r') as file:
    cyclists = pkl.load(file)
# Features
features = ['avg_position_profile_1', 'avg_position_profile_2','avg_position_profile_3', 'avg_position_profile_4', 'avg_position_profile_5']
ut.clustering_Agglomerate(
    cyclists,
    features,
    max_k=10,  # Maximum number of clusters for evaluation
    title='Cyclist',  # Title for the plots
    normalize_type='standard',  # Standard normalization
    K=3  # Predefined number of clusters
)
```
